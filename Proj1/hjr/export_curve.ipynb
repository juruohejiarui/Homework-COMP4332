```python
import tensorboard
from tensorboard.backend.event_processing import event_accumulator
import matplotlib.pyplot as plt

raw_datas = event_accumulator.EventAccumulator("./lora-qwen3-4b-l/runs/Mar19_19-30-20_hjr-MS-7D42/events.out.tfevents.1773919820.hjr-MS-7D42.129491.0")

raw_datas.Reload()
raw_datas.scalars.Keys()
```

```python
# make 3 curves: 1. train/macro_f1, 2. train/train_loss & eval/loss to mathplotlib using one figure with two y coordinate axes
train_loss = raw_datas.scalars.Items("train/loss")
eval_loss = raw_datas.scalars.Items("eval/loss")
eval_f1 = raw_datas.scalars.Items("eval/macro_f1")

train_loss_x = [x.step for x in train_loss]
train_loss_y = [x.value for x in train_loss]
eval_loss_x = [x.step for x in eval_loss]
eval_loss_y = [x.value for x in eval_loss]
eval_f1_x = [x.step for x in eval_f1]
eval_f1_y = [x.value for x in eval_f1]

fig, ax1 = plt.subplots(dpi=300)
ax2 = ax1.twinx()
ax1.plot(train_loss_x, train_loss_y, label="train_loss", color="blue")
ax1.plot(eval_loss_x, eval_loss_y, label="eval_loss", color="orange")
ax2.plot(eval_f1_x, eval_f1_y, label="eval_f1", color="green")
ax1.set_xlabel("step")
ax1.set_ylabel("loss")
ax1.set_ylim(0.5, 1.25)
ax2.set_ylabel("f1")
ax2.set_ylim(0.4, 0.72)
ax1.legend(loc="upper left")
ax2.legend(loc="upper right")
plt.title("train_loss, eval_loss and eval_f1")
plt.show()
```
